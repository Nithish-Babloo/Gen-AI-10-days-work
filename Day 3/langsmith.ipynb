```python
import os
from getpass import getpass

os.environ["GOOGLE_API_KEY"] = getpass("Enter Gemini API key: ")

os.environ["LANGSMITH_API_KEY"] = getpass("Enter LangSmith API key: ")

os.environ["LANGSMITH_TRACING"] = "true"
os.environ["LANGSMITH_PROJECT"] = "Gemini Demo"
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",

)
```

```python
from langsmith import traceable

@traceable
def generate_answer(question):
    response = model.invoke(question)
    return response.content
```

```python
answer = generate_answer("What is Artificial Intelligence?")
print(answer)
```
