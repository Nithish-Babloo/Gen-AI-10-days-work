```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("Aries/T5_question_answering")

model = AutoModelForSeq2SeqLM.from_pretrained(
    "Aries/T5_question_answering"
)
```

```python
context = """
Artificial Intelligence (AI) is a branch of computer science.
It enables machines to perform tasks that normally require
human intelligence, such as learning, reasoning, and problem solving.
"""

question = "What is AI?"

input_text = "question: " + question + " context: " + context

inputs = tokenizer(
    input_text,
    return_tensors="pt",
    truncation=True
)

outputs = model.generate(**inputs, max_length=50)

answer = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Answer:", answer)
```
