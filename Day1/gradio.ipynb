```python
from transformers import pipeline

generator = pipeline(
    "text-generation",
    model="gpt2"
)
```

```python
import gradio as gr

def generated_text(prompt):

    result = generator(
        prompt,
        max_new_tokens=100,
        do_sample=True,
        temperature=0.7
    )

    return result[0]["generated_text"]


interface = gr.Interface(
    fn=generated_text,
    inputs=gr.Textbox(
        lines=2,
        placeholder="Enter your prompt here"
    ),
    outputs="text",
    title="Text Generator"
)

interface.launch(share=True)
```
