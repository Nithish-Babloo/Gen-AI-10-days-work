```python
from google.colab import userdata
import os

os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

from langchain_core.messages import SystemMessage, HumanMessage

model = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0.7
)
```

```python
system = SystemMessage(content="""
You are a Food Question Answering Chatbot.

Answer only food-related questions.

Rules:
- Give simple and short answers.
- Use easy English.
- Keep answers within 3 to 5 sentences.
- Avoid unnecessary details.

For any question unrelated to food, reply exactly:

No, I can't answer to this at this time.
""")

user = "How many liters of water do we need to drink per day?"

response = model.invoke([
    system,
    HumanMessage(content=user)
])

print(response.content)
```

```python
user = "What are the benefits of eating fruits?"

response = model.invoke([
    system,
    HumanMessage(content=user)
])

print(response.content)
```
