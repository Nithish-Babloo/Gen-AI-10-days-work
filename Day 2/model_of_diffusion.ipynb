```python
from diffusers import StableDiffusionPipeline
import torch #frameowrk

model_id="sd-legacy/stable-diffusion-v1-5" #pretrain model
pipe=StableDiffusionPipeline.from_pretrained(model_id)
pipe=pipe.to("cuda")

prompt="a photograph of an astronanat riding a horse"
image=pipe(prompt).images[0]
display(image)
```
